```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn import svm
from sklearn.metrics import accuracy_score
import pickle
```

```python
# loading the diabetes dataset to a pandas DataFrame
diabetes_dataset = pd.read_csv('diabetes.csv')

# separating the data and labels
X = diabetes_dataset.drop(columns = 'Outcome', axis=1)
Y = diabetes_dataset['Outcome']
```

```python
# Split data FIRST before scaling to prevent data leakage
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, stratify=Y, random_state=2)
print(X.shape, X_train.shape, X_test.shape)
```

```python
# --- ADDED: Standardization Step ---
scaler = StandardScaler()

# Fit the scaler on the training data and transform it
X_train_scaled = scaler.fit_transform(X_train)

# Transform the test data using the SAME scaler (do not use fit_transform here!)
X_test_scaled = scaler.transform(X_test)
```

```python
# --- Train Classifier on SCALED Data ---
classifier = svm.SVC(kernel='linear')
classifier.fit(X_train_scaled, Y_train)

# accuracy score on the training data
X_train_prediction = classifier.predict(X_train_scaled)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
print('Accuracy score of the training data : ', training_data_accuracy)

# accuracy score on the test data
X_test_prediction = classifier.predict(X_test_scaled)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
print('Accuracy score of the test data : ', test_data_accuracy)
```

```python
# --- NEW: Save BOTH the trained classifier and the fitted scaler ---
model_filename = 'trained_model.sav'
scaler_filename = 'scaler.sav'

pickle.dump(classifier, open(model_filename, 'wb'))
pickle.dump(scaler, open(scaler_filename, 'wb'))
print("Model and Scaler successfully saved!")
```

```python
# ==============================================================
# --- VERIFICATION STEP (Simulating what your Web App will do) ---
# ==============================================================
loaded_model = pickle.load(open('trained_model.sav', 'rb'))
loaded_scaler = pickle.load(open('scaler.sav', 'rb'))

input_data = (5, 166, 72, 19, 175, 25.8, 0.587, 51)

# 1. Convert to numpy array
input_data_as_numpy_array = np.asarray(input_data)
# 2. Reshape for a single prediction instance
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)
# 3. Scale the input data using our loaded scaler before predicting
input_data_scaled = loaded_scaler.transform(input_data_reshaped)

# 4. Predict using the standardized data
prediction = loaded_model.predict(input_data_scaled)
print("Verification Prediction Code:", prediction)

if (prediction[0] == 0):
    print('The person is not diabetic')
else:
    print('The person is diabetic')
```
